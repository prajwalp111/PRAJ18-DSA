## if statemet



```python
if x > 2:
  print("the number-x is greater than 2")
```

```python
if x > 2 and y > 2:
  print("both number-x and y is greater than 2")
```

## if - else

```python
if x > 20 :
  print("x is greater than 20")
else:
  print("x is less than 20")
```

## if - elif - else

```python

if x < 0 and y < 0:
    print('X and Y is a -ve number')
elif x > 0 and y > 0:
    print('x and y is a +ve number')
else:
    print('str1 is definitely a string')
```

Ternary operator

code if condition else code

```python
print("1st condition passed") if x > 20 else print("2nd condition passed")
```

## Nested if

```python
a = 0
if a > 0:
    if a % 2 == 0:
        print('A is a positive and even integer')
    else:
        print('A is a positive number')
elif a == 0:
    print('A is zero')
else:
    print('A is a negative number')
```

## Match case


```python
status = 403

match status:
    case 401 | 403 | 404:
        print("Client error")

    case 500 | 502:
        print("Server error")
```
