## while statement

while - else

```python
count = 0
while count < 5:
    print(count)
    count = count + 1
else:
    print(f"else {count}")
```

break n continue



when break used else wont work

```python
count = 0
while count < 5:
    print(count)
    count = count + 1
    if count == 3:
        break
```

```python
count = 0
while count < 5:
    if count == 3:
        count += 1
        continue
    print(count)
    count = count + 1
```

## For loop


```python
numbers = [0, 1, 2, 3, 4, 5]
for number in numbers:
    print(number)
```

```python
language = 'Python' # string

for letter in language:
    print(letter)

print("----------------------")

for i in range(len(language)):
    print(language[i])
```

```python
numbers = (0, 1, 2, 3, 4, 5) # tuple
for number in numbers:
    print(number)
```

break n continue


```python
numbers = (0,1,2,3,4,5)
for number in numbers:
    print(number)
    if number == 3:
        break
```

```python
numbers = (0,1,2,3,4,5)
for number in numbers:
    print(number)
    if number == 3:
        continue
    print('Next number should be ', number + 1) if number != 5 else print("loop's end") # for short hand conditions need both if and else statements
print('outside the loop')
```

## Range functions


```python
lst = list(range(11))
print(lst)
st = set(range(1, 11))
print(st)

lst = list(range(0,11,2))
print(lst)
st = set(range(0,11,2))
print(st)

# for backward from start to end
lst = list(range(11,0,-2))
print(lst)
```

## Nested For Loop

```python
for key, value in person.items():
  if key == 'skills':
    for skills in value:
      print(skills)
```

## For Else

If we want to execute some message when the loop ends, we use else.



```python
for number in range(11):
    print(number)   # prints 0 to 10, not including 11
else:
    print('The loop stops at', number)
```
