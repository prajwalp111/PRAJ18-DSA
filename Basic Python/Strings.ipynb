## Strings

```python
str1 = "alphabets"
str2 = "apple - fruits"

print(f"{str1} and {str2}")
```

```python
str_multiline = """"This is a multi-line string. 
It can span multiple lines and --   -- preserve the formatting of the test.
this 2nd line"""
print(str_multiline)
```

### Accessing Characters in String

```python
print(f"{str1[0]} this prints the 1st character of str1") 
print(f"{str1[5]} this prints the 6th character of str1")       
```

Reading string -ve indexing


```python
print(f"{str2[-1]} this prints the last character of str2")
print(f"{str2[-5]} this prints the 5th last character of str2")
```

Slicing

```python
print(f"{str2[:4]:10} - this prints the first 4 characters of str2")

print(f"{str2[6:]:10} - this prints the characters of str2 from index 6 to the end")

print(f"{str2[::2]:10} - this prints every 2nd character of str2")

print(f"{str2[::-1]:20} - this prints the characters of str2 in reverse order")

#{str[..]:a}   here the number that tell to give it a minimum width of 10 characters
```

Looping Through Strings


```python
str3 = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
```

```python
for i in str3:
    print(f"{i} - {str3.index(i)}")
```

```python
for i in range(0, len(str3), 5):
    print(f"{str3[i:i+5]:10} - this prints 5 characters of str3 starting from index {i}")
```

### String Immutability 
Strings are immutable, meaning their values cannot be changed after creation. Any modification to a string creates a new string instead of altering the original one

```python
s = "aBCDEF"
s = "A" + s[1:]  
print(s)
```

Deleting a String

```python
s = "ABC"
del s
```
