#Sets

Sets are unordered and mutable collections used to store unique elements.


```python
s = {'g', 'a', 't', 'e'}
print(s[0]) # can't access sets using the indexing
```

```python
odd_nums = range(1, 50, 2)
nums = set(odd_nums)

for n in nums :
  print(n)
```

```python
animals = {"dog", 'cat', 'deer', 'squirrel'}

for i, a in enumerate(animals):
  print(f"{i} => {a} ")
```

```python
A = {1, 2}
B = {2, 1, 2}
C = {1, 2, 3}
D = {2}

# Equality Check
print(f"Equality Check: {A == B}")
# True

# Proper Subset Check
print(f"Proper Subset Check: {A < C}")
# True

# Subset Check
print(f"Subset Check: {A <= C}")
# True

# Proper Superset Check
print(f"Proper Superset Check: {A > D}")
# True

# Superset Check
print(f"Superset Check: {A >= A}")
# True
```

```python
s = {1, 2}
s.add(3)
print(s)
```

```python
s = {1, 2}
s.update([3, 4])
print(s)

s.update("AB")
print(s)
```

```python
s = {1, 2, 3}
s.remove(2)
print(s)
print(s.remove(6))
```

```python
s = {1, 2, 3}
s.discard(5)
# no error, even though 5 is absent
print(s)
```

```python
s = {10, 20, 30}
elem = s.pop()
print(elem, s)
```

```python
s = {1, 2, 3}
s.clear()
print(s)
```

```python
nums = {4, 2, 7}
nums2 = sorted(nums)
print(nums2)

# a list, not a set

nums2 = sorted(nums, reverse=True)
print(nums2)
```
