#Dictionary

Dictionary is a data structure that stores information in key-value pairs.

  

* While keys must be unique and immutable (like strings or numbers), values can be of any data type, whether mutable or immutable.
*   Dictionaries ideal for accessing data by a specific name rather than a numeric position like in list.

```python
s = {}
print(s, type(s))
```

```python
s = {1,2,3,4}
print(s, type(s))
```

```python
person = {'name': 'John', 'age': 30}
print(person, type(person))
```

```python
person = dict(name="Bhishma", age=999 )
print(person)

# using the tuple
items = dict([('a',1), ('b',2), ('c',3)]) # [ (), () ] imp
print(items)
```

```python
keys = ['a', 'b', 'c']
new_dict = dict.fromkeys(keys, 0)
print(new_dict)


d = dict.fromkeys(['a', 'b', 'c'], [])
"""If the default value is MUTABLE (like []),
every key gets bound to the exact SAME object"""
d['a'].append(1)
print(d)
```

```python
student = {"name": "Alice", "age": 22}
print(student["name"])
print(student["grade"])
```

```python
student = {"name": "Alice", "age": 22}

print(student.get("name"))

print(student.get("grade"))
"""returns default_value (or None if omitted) instead of raising an
 error when the key is missing"""
print(student.get("grade", "N/A"))
```

```python
d = {'a': 1}
print(1 in d)         # False   — 1 is a VALUE, not a key
print('a' in d)       # True    — 'a' IS a key
print(1 in d.values()) # True    — explicitly check values instead
```

```python
student = {"name": "Alice", "age": 22, "grade": "A"}
print(student.keys())

print(student.values())

print(student.items())
```

```python
d = {'a': 1, 'b': 2, 'c': 3}
print(len(d))
```

```python
student = {"name": "Alice", "age": 22}
for key in student:
    print(key, student[key])

for value in student.values():
    print(value)

for key, value in student.items():
    print(key, "->", value)
```

```python
student = {"name": "Alice", "age": 21}
student["grade"] = "A"      # adds a NEW key
student["age"] = 22          # UPDATES an existing key
print(student)
```

```python
student = {"name": "Alice", "age": 21}
student.update({"age": 23, "gender": "Female"})
print(student)
```

```python
data = {"a": 1, "b": 2, "c": 3}
value = data.pop("c")
print(value)    # 3
print(data)     # {'a': 1, 'b': 2}

#print(del data['b'])

#data.pop("z")
# Error: KeyError: 'z'
data = {"x": 10, "y": 20}
item = data.popitem()
print(item)     # ('y', 20)   — the LAST-inserted pair

#{}.popitem()
# Error: KeyError: 'popitem(): dictionary is empty'
```

```python
student = {
'name': 'Alice',
'marks': {'math': 85, 'science': 90},
'address': {'city': 'Hyderabad', 'pincode': 500001}
}
print(student['marks']['math'])
print(student['address']['city'])
```

copy


```python
d1 = {'a': 1, 'b': [2, 3]}
d2 = d1
d2['a'] = 2
d2['b'].append(4)
print(d2)
print(d1)
```

```python
d1 = {'a': 1, 'b': [2, 3]}
d2 = d1.copy()
d2['a'] = 2
   # only affects d2 — 'a' is an immutable int, no sharing possible
d2['b'].append(4)
   # affects d1 TOO — 'b' is a shared list
print(f'd1:{d1}')
print(f'd2:{d2}')
```

```python
import copy
d1 = {'a': 1, 'b': [2, 3]}
d2 = copy.deepcopy(d1)
d2['b'].append(4)

 # {'a': 1, 'b': [2, 3]}       — completely unaffected
print(f'd1:{d1}')
print(f'd2:{d2}')
```
