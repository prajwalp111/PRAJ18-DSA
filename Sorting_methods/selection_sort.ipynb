# Sorting methods


## Selection Sort

- **Time Complexity:** `O(n²)`
  - Best Case: `O(n²)`
  - Average Case: `O(n²)`
  - Worst Case: `O(n²)`

- **Space Complexity:** `O(1)` — In-place sorting

```python
L = [-3,10, 9, -2,1,-3,6,5]
```

```python
def selection_sort(arr):
    for i in range(len(arr)):
        min = i
        for j in range(i+1, len(arr)):
            if arr[j] < arr[min]:
                min = j
        arr[i], arr[min] = arr[min], arr[i]
    return arr


selection_sort(L)
```

## Bubble Sort

- **Time Complexity:** `O(n²)`
  - Best Case: `O(n)` — with optimization/check for no swaps
  - Average Case: `O(n²)`
  - Worst Case: `O(n²)`

- **Space Complexity:** `O(1)` — In-place sorting

```python
def bubble_sort(arr):
    for i in range(len(arr)-1,0,-1):
        for j in range(i):
            if arr[j] > arr[j+1]:
              arr[j], arr[j+1] = arr[j+1], arr[j]

    return arr

bubble_sort(L)
```
